```python
#multi inheritance
class father:
    def __init__(self,fname,profession):
        self.fname = fname
        self.profession=profession
        
    def display(self):
        print("family details")
        print("father name:",self.fname)
        print("father job",self.profession)
        
class mother:
    def __init__(self,mname,job):
        self.mname=mname
        self.job=job
    def output(self):
        print("mother name:",self.mname)
        print("job:",self.job)
class son(father,mother):
    def __init__(self,fname,profession,mname,job,name,standard):
        father.__init__(self,fname,profession)
        mother.__init__(self,mname,job)
        self.name=name
        self.standard=standard
    def dlays(self):
        father.display(self)
        mother.output(self)
        print("name:",self.name)
        print("standard:",self.standard)
s=son("netteppa","JSW","thirpal","housewife","mahi",16)

s.dlays()
```

```python
#sorting algorithm
#bubble sort
a=[9,3,4,5,2,0,1]
for i in range(len(a)):
    for j in range(0,len(a)-i-1):
        if a[j]>a[j+1]:
            temp=a[j]
            a[j]=a[j+1]
            a[j+1]=temp
print("sorted array",a)
```

```python
#merge sort

def merges(a):
    if len(a)>1:
        mid=len(a)//2
        left=a[:mid]
        right=a[mid:]
        merges(left)
        merges(right)
        i=j=k=0
        while i<len(left) and j<len(right):
            if left[i] < right[j]:
                a[k]=left[i]
                i+=1       
            else:
                a[k]=right[j]
                j+=1
            k+=1
        while i<len(left):
            a[k]=left[i]
            i+=1
            k+=1
    
        while j<len(right):
            a[k]=right[j]
            j+=1
            k+=1

    
a=list(map(int,input()))
merges(a)
print("sorted array",a)
```

```python
def partition(arr,low,high):
    pivot=arr[high]
    i=low-1
    for j in range(low,high):
        if arr[j]<=pivot:
            i+=1
            arr[i],arr[j]=arr[j],arr[i]
        
    arr[i+1],arr[high]=arr[high],arr[i+1]
    return i+1
def quicksort(arr,low,high):
    if low<high:
        pivot_index=partition(arr,low,high)
        quicksort(arr,low,pivot_index-1)
        quicksort(arr,pivot_index+1,high)
        
arr=list(map(int,input()))
quicksort(arr,0,len(arr)-1)
print("sorted array",arr)
```

```python
# Singly Linked List - Insert and Delete

# Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Linked List class
class SinglyLinkedList:
    def __init__(self):
        self.head = None

    # Insert at beginning
    def insert_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # Insert at end
    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        temp = self.head

        while temp.next is not None:
            temp = temp.next

        temp.next = new_node

    # Insert after a given value
    def insert_after(self, value, data):
        temp = self.head

        while temp is not None:
            if temp.data == value:
                new_node = Node(data)
                new_node.next = temp.next
                temp.next = new_node
                return

            temp = temp.next

        print("Value", value, "not found.")

    # Delete a node by value
    def delete(self, value):
        if self.head is None:
            print("List is empty.")
            return

        # If the first node is the node to delete
        if self.head.data == value:
            self.head = self.head.next
            return

        temp = self.head

        while temp.next is not None:
            if temp.next.data == value:
                temp.next = temp.next.next
                return

            temp = temp.next

        print("Value", value, "not found.")

    # Display the linked list
    def display(self):
        temp = self.head

        while temp is not None:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("NULL")


# -----------------------------
# Main Program
# -----------------------------

linked_list = SinglyLinkedList()

# Initial list: 10, 20, 30
linked_list.insert_end(10)
linked_list.insert_end(20)
linked_list.insert_end(30)

print("Initial List:")
linked_list.display()

# Insert 5 at beginning
linked_list.insert_beginning(5)

# Insert 40 at end
linked_list.insert_end(40)

# Insert 25 after 20
linked_list.insert_after(20, 25)

# Delete 30
linked_list.delete(30)

print("\nFinal List:")
linked_list.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next is not None:
            temp = temp.next

        temp.next = new_node

    def reverse(self):
        previous = None
        current = self.head

        while current is not None:
            next_node = current.next
            current.next = previous
            previous = current
            current = next_node

        self.head = previous

    def display(self):
        temp = self.head

        while temp is not None:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("NULL")


linked_list = SinglyLinkedList()

linked_list.insert_end(10)
linked_list.insert_end(20)
linked_list.insert_end(30)
linked_list.insert_end(40)

print("Original:")
linked_list.display()

linked_list.reverse()

print("After Reverse:")
linked_list.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_beginning(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    def insert_end(self, data):
        new_node = Node(data)

        if self.tail is None:
            self.head = self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

    def insert_before(self, value, data):
        current = self.head

        while current is not None:
            if current.data == value:
                new_node = Node(data)

                if current == self.head:
                    new_node.next = current
                    current.prev = new_node
                    self.head = new_node
                else:
                    new_node.prev = current.prev
                    new_node.next = current
                    current.prev.next = new_node
                    current.prev = new_node

                return

            current = current.next

        print("Value not found.")

    def insert_after(self, value, data):
        current = self.head

        while current is not None:
            if current.data == value:
                new_node = Node(data)

                new_node.prev = current
                new_node.next = current.next

                if current.next is not None:
                    current.next.prev = new_node
                else:
                    self.tail = new_node

                current.next = new_node
                return

            current = current.next

        print("Value not found.")

    def delete(self, value):
        current = self.head

        while current is not None:
            if current.data == value:
                if current.prev is not None:
                    current.prev.next = current.next
                else:
                    self.head = current.next

                if current.next is not None:
                    current.next.prev = current.prev
                else:
                    self.tail = current.prev

                return

            current = current.next

        print("Value not found.")

    def search(self, value):
        current = self.head

        while current is not None:
            if current.data == value:
                return True
            current = current.next

        return False

    def display_forward(self):
        current = self.head

        while current is not None:
            print(current.data, end=" <-> ")
            current = current.next

        print("NULL")

    def display_backward(self):
        current = self.tail

        while current is not None:
            print(current.data, end=" <-> ")
            current = current.prev

        print("NULL")


dll = DoublyLinkedList()

dll.insert_end(10)
dll.insert_end(20)
dll.insert_end(30)
dll.insert_end(40)

dll.insert_beginning(5)
dll.insert_before(20, 15)
dll.insert_after(20, 25)

dll.delete(5)
dll.delete(10)

print("Forward:")
dll.display_forward()

print("Backward:")
dll.display_backward()

print("Search 30:", dll.search(30))
print("Search 50:", dll.search(50))
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self, n):
        self.head = None
        self.tail = None

        for i in range(1, n + 1):
            self.insert(i)

    def insert(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
            new_node.next = self.head
        else:
            new_node.next = self.head
            self.tail.next = new_node
            self.tail = new_node

    def remove_kth(self, k):
        current = self.head
        previous = self.tail

        while current.next != current:
            for _ in range(k - 1):
                previous = current
                current = current.next

            previous.next = current.next

            if current == self.head:
                self.head = current.next

            current = previous.next

        self.head = current
        self.tail = previous

        return self.head.data


n = 7
k = 3

circular_list = CircularLinkedList(n)
result = circular_list.remove_kth(k)

print("Remaining Node =", result)
```

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        return None

    def is_empty(self):
        return len(self.stack) == 0


def is_balanced(expression):
    stack = Stack()

    pairs = {
        ')': '(',
        '}': '{',
        ']': '['
    }

    for char in expression:
        if char in "({[":
            stack.push(char)
        elif char in ")}]":
            top = stack.pop()

            if top != pairs[char]:
                return False

    return stack.is_empty()


expression = input("Input: ")

if is_balanced(expression):
    print("Output: Balanced Parentheses")
else:
    print("Output: Not Balanced Parentheses")
```

```python
class Queue:
    def __init__(self):
        self.queue = []

    def add_customer(self, name):
        self.queue.append(name)

    def serve_customer(self):
        if self.is_empty():
            print("No customers waiting.")
        else:
            print("Served Customer:", self.queue.pop(0))

    def show_customers(self):
        if self.is_empty():
            print("No customers waiting.")
        else:
            print("Waiting Customers:", " -> ".join(self.queue))

    def next_customer(self):
        if self.is_empty():
            print("No customers waiting.")
        else:
            print("Next Customer:", self.queue[0])

    def is_empty(self):
        return len(self.queue) == 0


queue = Queue()

while True:
    print("\n1. Add Customer")
    print("2. Serve Customer")
    print("3. Show Customers")
    print("4. Next Customer")
    print("5. Waiting Customers")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter Customer Name: ")
        queue.add_customer(name)
        print("Customer Added:", name)

    elif choice == "2":
        queue.serve_customer()

    elif choice == "3":
        queue.show_customers()

    elif choice == "4":
        queue.next_customer()

    elif choice == "5":
        queue.show_customers()

    elif choice == "6":
        print("Thank you!")
        break

    else:
        print("Invalid choice.")
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    def insert(self, data):
        new_node = Node(data)

        if self.root is None:
            self.root = new_node
            return

        current = self.root

        while True:
            if data < current.data:
                if current.left is None:
                    current.left = new_node
                    return
                current = current.left
            else:
                if current.right is None:
                    current.right = new_node
                    return
                current = current.right

    def inorder(self, node, result):
        if node is not None:
            self.inorder(node.left, result)
            result.append(node.data)
            self.inorder(node.right, result)

    def preorder(self, node, result):
        if node is not None:
            result.append(node.data)
            self.preorder(node.left, result)
            self.preorder(node.right, result)

    def postorder(self, node, result):
        if node is not None:
            self.postorder(node.left, result)
            self.postorder(node.right, result)
            result.append(node.data)

    def count_nodes(self, node):
        if node is None:
            return 0
        return 1 + self.count_nodes(node.left) + self.count_nodes(node.right)

    def count_leaf_nodes(self, node):
        if node is None:
            return 0

        if node.left is None and node.right is None:
            return 1

        return self.count_leaf_nodes(node.left) + self.count_leaf_nodes(node.right)

    def height(self, node):
        if node is None:
            return 0

        left_height = self.height(node.left)
        right_height = self.height(node.right)

        return 1 + max(left_height, right_height)


tree = BinaryTree()

values = [50, 30, 70, 20, 40, 60, 80, 10]

for value in values:
    tree.insert(value)

inorder_result = []
preorder_result = []
postorder_result = []

tree.inorder(tree.root, inorder_result)
tree.preorder(tree.root, preorder_result)
tree.postorder(tree.root, postorder_result)

print("Inorder:", *inorder_result)
print("Preorder:", *preorder_result)
print("Postorder:", *postorder_result)
print("Total Nodes:", tree.count_nodes(tree.root))
print("Leaf Nodes:", tree.count_leaf_nodes(tree.root))
print("Height:", tree.height(tree.root))
```
